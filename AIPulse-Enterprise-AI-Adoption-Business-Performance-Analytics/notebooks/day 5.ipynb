
### Decide keep / drop / impute rule for each affected column





### 1. Missing Values

**Rule:**

* If a column has **0 missing values**, **keep** it as it is.
* If missing values are found in the future:

  * Numeric columns → Impute with **mean** or **median**.
  * Categorical columns → Impute with **mode**.
  * If a column has too many missing values (e.g., >50%), consider dropping it after evaluating its importance.

**finding:**

> No missing values were found, so all columns are kept and no imputation is required.

---

### 2. Duplicate Rows

**Rule:**

* If duplicate rows are found, remove them using `drop_duplicates()`.
* If no duplicates exist, keep all records.

**Your finding:**

> No duplicate rows were found, so no rows were dropped.

---

### 3. Data Type Fixes

**Rule:**

* If a numeric column is stored as text, convert it to a numeric type using `pd.to_numeric()`.
* If date columns are stored as text, convert them using `pd.to_datetime()`.
* If data types are already correct, keep them unchanged.

**finding:**

> All columns have correct data types, so no conversion is required.

---

### 4. Category Harmonisation

**Rule:**

* Remove leading and trailing spaces using `.str.strip()`.
* Standardize text case if needed using `.str.lower()` or `.str.title()`.
* Correct inconsistent spellings if they exist.

**finding:**

> Categories were already consistent. Only whitespace was checked, and no changes were needed.

---

## Cleaning Plan Summary

| Issue                  | Rule                              |  Decision |
| ---------------------- | --------------------------------- | ------------- |
| Missing Values         | Impute if present; otherwise keep | **Keep**      |
| Duplicate Rows         | Drop duplicates if found          | **Keep**      |
| Data Types             | Convert incorrect types           | **Keep**      |
| Category Harmonisation | Standardize spaces and case       | **Keep**      |
