#### Task 2: Apply Cleaning Plan Using Pandas

```python
company_clean["region"] = company_clean["region"].str.strip()
company_clean["survey_source"] = company_clean["survey_source"].str.strip()
company_clean[["region", "survey_source"]].head()
```

```python
industry_clean["industry"] = industry_clean["industry"].str.strip()
```

```python
country_clean["country"] = country_clean["country"].str.strip()
country_clean["region"] = country_clean["region"].str.strip()
```

### Harmonise Categories

```python
company_clean["region"] = company_clean["region"].str.title()
company_clean["survey_source"] = company_clean["survey_source"].str.title()
```

```python
industry_clean["industry"] = industry_clean["industry"].str.title()
```

```python
country_clean["country"] = country_clean["country"].str.title()
country_clean["region"] = country_clean["region"].str.title()
```
