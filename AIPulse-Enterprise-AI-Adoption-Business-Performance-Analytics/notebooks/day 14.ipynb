#### Task 2

#### Distribution analysis (Histogram, Skewness, Kurtosis) for:

#### ai_adoption_rate
#### productivity_change_percent
### ai_maturity_score

Histogram

A histogram shows how the values are distributed. It helps identify where most observations are concentrated and whether the data is symmetric or skewed.

Skewness

Skewness measures the asymmetry of the distribution.

Skewness ≈ 0 → Symmetric distribution
Skewness > 0 → Right-skewed (long tail on the right)
Skewness < 0 → Left-skewed (long tail on the left)
Kurtosis

Kurtosis measures how heavy the tails of the distribution are.

Kurtosis > 0 → More extreme values (heavy tails)
Kurtosis ≈ 0 → Similar to a normal distribution
Kurtosis < 0 → Fewer extreme values (light tails)

```python
columns = [
    "ai_adoption_rate",
    "productivity_change_percent",
    "ai_maturity_score"
]

for column in columns:

    plt.figure(figsize=(8,5))

    plt.hist(
        df_company[column],
        bins=25,
        edgecolor="black"
    )

    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")

    plt.grid(alpha=0.3)

    plt.show()
```

```python
### skewness
for column in columns:

    print("="*60)
    print("Column :", column)

    print("Skewness :", round(df_company[column].skew(),3))
    print("Kurtosis :", round(df_company[column].kurt(),3))
```

```python
for column in columns:

    skew = df_company[column].skew()
    kurt = df_company[column].kurt()

    print("="*60)
    print("Column :", column)

    print("Skewness :", round(skew,3))
    print("Kurtosis :", round(kurt,3))

    if skew > 0:
        print("Distribution : Positively Skewed")
    elif skew < 0:
        print("Distribution : Negatively Skewed")
    else:
        print("Distribution : Symmetric")

    if kurt > 0:
        print("Tail : Heavy tails (more extreme values)")
    elif kurt < 0:
        print("Tail : Light tails (fewer extreme values)")
    else:
        print("Tail : Normal tails")
```

### Observation:
-- The histogram analysis shows the distribution of AI adoption rate, productivity change percentage, and AI maturity score across companies. 
   Skewness helps identify whether the data is symmetric or biased toward higher or lower values, while kurtosis indicates the presence of extreme values. 
   These statistics improve the understanding of data distribution and support appropriate analytical decisions.
